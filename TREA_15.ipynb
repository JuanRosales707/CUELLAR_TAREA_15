```python
# Importar librerías necesarias
import chromadb
from chromadb.config import Settings
from sentence_transformers import SentenceTransformer
```

```python
# Crear la base de datos local y colección
client = chromadb.PersistentClient(path="./db")
collection = client.get_or_create_collection(name="Tecsup")
```

```python
# Información personalizada (Gerbert y amigos)
documentos = [
    "Mi nombre es Gerbert. Tengo 18 años y estudio Big Data y Ciencia de Datos en Lima. Me apasiona mantenerme activo y explorar nuevas ideas. Suelo decir: 'Siempre hay espacio para aprender más'.",
    "Sofía, tiene 20 años y también estudia Big Data y Ciencia de Datos en Lima. Disfruta descubrir nuevas perspectivas mientras descanso. Mi lema es: 'Cada día trae su propio aprendizaje'.",
    "Óscar, tiene 25 años. Estudia Big Data y Ciencia de Datos en Lima. Le gusta despejar la mente a través del ejercicio y actividades recreativas. Siempre digo: 'Nada mejor que avanzar paso a paso'.",
    "Julio. Tiene 19 años y es estudiante de Big Data y Ciencia de Datos en Lima. Encuentra placer en conectar ideas con personas. Suele decir: 'La clave está en compartir lo que sabes'."
]
# Agregar los documentos con IDs únicos
collection.add(
    documents=documentos,
    ids=["gerbert", "sofia", "oscar", "julio"]
)
```

```python
# Crear un prompt (consulta)
prompt = "¿Quién estudia en Lima y disfruta mantenerse activo?"
resultados = collection.query(
    query_texts=[prompt],
    n_results=2
)

# Mostrar resultados
print("Consulta:", prompt)
for r in resultados['documents'][0]:
    print("\nResultado:\n", r)
```

```python
# Visualizar lo que hay en la colección
print("\n--- Datos en la base vectorial 'Tecsup' ---")
print(collection.peek(4))  # mostrar los 4 documentos insertados
```

```python
# Comparación con otro motor de embeddings
from sentence_transformers import SentenceTransformer

# Crear nueva colección en el mismo cliente
mini_collection = client.get_or_create_collection(name="Tecsup_MiniLM")

# Cargar el modelo all-MiniLM-L6-v2
modelo_mini = SentenceTransformer('all-MiniLM-L6-v2')

# Convertir documentos en vectores
vectores_mini = modelo_mini.encode(documentos).tolist()

# Agregar a la nueva colección
mini_collection.add(
    documents=documentos,
    ids=["gerbert_m", "sofia_m", "oscar_m", "julio_m"],
    embeddings=vectores_mini
)

# Consultar con el nuevo motor de embeddings
prompt = "¿Quién estudia en Lima y disfruta mantenerse activo?"
vector_prompt = modelo_mini.encode([prompt]).tolist()

resultados_mini = mini_collection.query(
    query_embeddings=vector_prompt,
    n_results=2
)

# Mostrar resultados
print("\n--- Comparación usando el modelo 'all-MiniLM-L6-v2' ---")
print("Consulta:", prompt)
for r in resultados_mini['documents'][0]:
    print("\nResultado:\n", r)
```
